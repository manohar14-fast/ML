CODE FOR MULTIPE LINEAR RREGRESSON

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = {
    'Area': [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8],
    'Rooms': [2, 2, 3, 3, 4, 4, 5, 6],
    'Age': [10, 15, 18, 12, 14, 18, 16, 15],
    'Distance': [8, 4, 3, 7, 12, 4, 6, 13],
    'Price': [30, 35, 45, 58, 80, 60, 88, 99]
}
#AP24110012036
df = pd.DataFrame(data)
print(df)
```

```python
#TO GET THE I/P AND O/P:
X = df[['Area', 'Rooms', 'Age', 'Distance']]
y = df['Price']
```

```python
model = LinearRegression()
model.fit(X, y)
print("Model trained successfully")
```

```python
print("Coefficients:", np.round(model.coef_, 2))
print("Intercept:", round(model.intercept_, 2))
```

```python
#PREDICTING THE HOUSE PRICES

y_pred = model.predict(X)
  #AP24110012036
print("Actual Prices:", y.values)
print("Predicted Prices:", y_pred.round(2))
```

```python
#PREDICTING A NEW HOUSE PRICE
new_house = pd.DataFrame({
    'Area': [2.4],
    'Rooms': [4],
    'Age': [12],
    'Distance': [3]
})

price = model.predict(new_house)

print("Predicted House Price:", round(price[0], 2), "Lakhs")
```

```python
# TO CALCULATE R^2:

print("R2 Score:", round(model.score(X, y), 2))
```
