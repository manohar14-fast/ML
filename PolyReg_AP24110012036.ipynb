## ** POLYNOMIAL REGRESSION - AP24110012036**
CODE FOR POLYNOMIAL REGRESSION USING  HOUSE PRICE PREDICTION EXAMPLE


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
data = {
    'Area': [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8],
    'Price': [30, 35, 45, 58, 80, 60, 88, 99]
}
#AP24110012036
df = pd.DataFrame(data)
print(df)
```

```python
X = df[['Area']]
y = df['Price']  #AP24110012036
```

```python
#CONVERTING THE DATA INTO A PEOLYNOMIAL FEATURE

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

print(X_poly)
```

```python
model = LinearRegression()
model.fit(X_poly, y)

print("Model trained successfully")
```

```python
#DISPLAYING OUTPUTVALUES
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
#FOR THE SAKE OF PREDICTING THE HOUSE'S PRICES:

y_pred = model.predict(X_poly)

print("Actual Prices:", y.values)
print("Predicted Prices:", y_pred.round(2))
```

```python
#PREDITING FOR THE NEW HOUSE

new_area = pd.DataFrame({'Area': [2.4]})

new_area_poly = poly.transform(new_area)
price = model.predict(new_area_poly)

print("Predicted House Price:", round(price[0], 2), "Lakhs")
```

```python
#TO PLOT THE CURVE
plt.scatter(X, y, color='blue', label='Actual Prices')

plt.plot(X, y_pred, color='red', label='Polynomial Curve')

plt.xlabel("Area")
plt.ylabel("Price (Lakhs)")
plt.title("Polynomial Regression")  #AP24110010230
plt.legend()
plt.show()
```

```python
#TO GET THE R^2 VALUE WE USE :
print("R2 Score:", round(model.score(X_poly, y), 2))
```
