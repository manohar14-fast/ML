BASIC CODE FOR THE LINEAR RREGRESSION :

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#CREATING THE DATASET FOR THE CODE - USING THE HOUSE RENT AS AN EXAMPLE FOR IT
data = {
    'Area': [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8],
    'Price': [32,40 , 45, 58, 86, 90, 92, 95]
}

df = pd.DataFrame(data)
print(df)  #AP24110012036

X = df[['Area']]
y = df['Price']
```

```python
#TO TRAIN THE MODEL WE USE
model = LinearRegression()
model.fit(X, y)

print("Model trained successfully")
```

```python
print("Slope:", round(model.coef_[0], 2))
print("Intercept:", round(model.intercept_, 2))
```

```python
# TO PREDICT THE PRICES
y_pred = model.predict(X)

print("Actual Prices:", y.values)
print("Predicted Prices:", y_pred.round(2))
```

```python
#PREDICT A NEW HOUSE PRICES

new_area = pd.DataFrame({'Area': [2.4]})
price = model.predict(new_area)
print("Predicted House Price:", round(price[0], 2), "Lakhs")
```

```python
# to get the regression line

plt.scatter(X, y, color='blue', label='Actual Prices')
plt.plot(X, y_pred, color='red', label='Regression Line')

plt.xlabel("Area")
plt.ylabel("Price (Lakhs)")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```
